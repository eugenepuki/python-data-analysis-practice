```python
# loop version
grades = [18, 25, 30, 14, 27, 21]

def maximum_grade(grades):
    maximum = grades[0]
    for grade in grades:
        if grade > maximum:
            maximum = grade
    return maximum

result = maximum_grade(grades)
print("maximum grade (loop version):", result)
```

```python
# Student Grade Analyzer

# Project 1 - Python Data Analysis Practice

grades = [18, 25, 30, 14, 27, 21]

def average_grade(grades):

    total = sum(grades)

    number_of_grades = len(grades)

    average = total / number_of_grades

    return average

def lowest_grade(grades):

    return min(grades)

def highest_grade(grades):

    return max(grades)

def count_passing_grades(grades):

    count = 0

    for grade in grades:

        if grade >= 18:

            count += 1

    return count

def filter_high_grades(grades):

    high_grades = []

    for grade in grades:

        if grade >= 25:

            high_grades.append(grade)

    return high_grades

# Display the results

print("Average grade:", average_grade(grades))

print("Maximum grade:", highest_grade(grades))

print("Lowest grade:", lowest_grade(grades))

print("Passing grades:", count_passing_grades(grades))

print("Grades of 25 or higher:", filter_high_grades(grades))
```
